```python
df.columns=df.columns.str.strip().str.lower().str.replace(' ','_').str.replace('-','')
```

Standardizing Names

```python
df['customer_name']=df['customer_name'].str.strip().str.title().str.replace(' ','')
```

Standardizing Email Address

```python
df['email_address']=df['email_address'].str.strip().str.lower().str.replace(' ','')
```

You can also Separate Email Search engine and UserName by (expand)

```python
df['email_address'].str.strip().str.lower().str.replace(' ','').str.split('@',expand=True)
```

```python
def registration_clean(date):
    if pd.isna(date):
        return np.nan
    date=str(date)
    # return np.nan

    split=date.split('/')
    if split[1]=='Jan':
        split[1]=='01'
```
