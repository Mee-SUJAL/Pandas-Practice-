HR → 5% increase
Finance → 12% increase
Sales → 15% increase]
Marketing → 8% increase

```python
df=pd.read_csv('Employee.csv')
```

```python
# Map department to increment percentage
salary_increase = {
    'IT': 1.10,
    'HR': 1.05,
    'Finance': 1.12,
    'Sales': 1.15,
    'Marketing': 1.08
}

# Create Adjusted_Salary column
df['Adjusted_Salary'] = df['Salary'] * (1 + df['Department'].map({
    'IT': 1.10,
    'HR': 1.05,
    'Finance': 1.12,
    'Sales': 1.15,
    'Marketing': 1.08
}
))
```
